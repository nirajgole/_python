# LinkedList

### Singly

```python
class Node:
    def __init__(self,data,next=None):
        # each node has data and next pointer refering to next node in list
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_start(self,data):
        node = Node(data,self.head)
        self.head = node

    def insert_at_end(self,data):
        if self.head is None:
            self.head = Node(data,None)
            return

        itr = self.head
        while itr.next:
            itr = itr.next
        itr.next = Node(data,None)

    def remove_at_start(self):
        if self.head is None:
            return

        temp = self.head.next
        self.head = temp
        del temp

    def remove_at_end(self):
        if self.head is None:
            return

        cur = self.head
        while cur:
            if cur.next.next is None:
                break
            cur = cur.next
        temp = cur.next
        cur.next = None
        del temp

    def insert_at(self,index:int,data:int):
        #try for negative index
        #index starting from 0
        _length = self.get_length()
        if index >= _length or index < 0:
            raise Exception("Index out of range")
        if index == 0:
            self.insert_at_start(data)
        else:
            node = Node(data)
            itr = self.head
            count = 1
            while itr:
                if count == index:
                    temp = itr.next
                    itr.next = node
                    node.next = temp
                    break
                count+=1
                itr = itr.next

    def remove_at(self,index:int):
        _length = self.get_length()
        if index >= _length or index < 0:
            raise Exception("Index out of range")
        if index == 0:
            self.remove_at_start()
        if index == _length-1:
            self.remove_at_end()
        else:
            itr = self.head
            count = 1
            while itr:
                if count == index:
                    temp = itr.next
                    itr.next = itr.next.next
                    del temp
                    break
                count+=1
                itr = itr.next


    def get_length(self):
        if self.head is None:
            return 0
        count = 0
        itr = self.head
        while itr:
            count += 1
            itr = itr.next
        return count

    def print_list(self):
        if self.head is None:
            print("List is empty")
            return

        itr = self.head
        while itr:
            print(itr.data, end=" -> ")
            itr = itr.next
```

### Doubly

```python
class Node:
    def __init__(self,data,prev=None,next=None):
        # each node has data and next pointer refering to next node in list
        self.prev = prev
        self.data = data
        self.next = next

class DoublyLinkedList:
    def __init__(self):
        self.head = None

    #push
    def insert_at_start(self,data):
        node = Node(data)
        if self.head is None:
            self.head = node
        else:
            self.head.prev = node
            node.next = self.head
            self.head = node
    #append
    def insert_at_end(self,data):
        node = Node(data)
        if self.head is None:
            self.head = node
        else:
            itr = self.head
            while itr.next:
                itr = itr.next
            itr.next = node
            node.prev = itr
    #pop
    def remove_at_start(self):
        if self.head is None:
            print("List is empty")
        else:
            self.head = self.head.next
            self.head.prev = None

    #remove last node
    def remove_at_end(self):
        if self.head is None:
            print("List is empty")
        else:
            itr = self.head
            while itr.next:
                itr = itr.next
            temp = itr.prev
            temp.next = None
            itr.prev = None

    def insert_at(self,index:int,data:int):
        itr = self.head
        count = 0
        if count == index:
            self.insert_at_start(data)
            return
        while itr.next:
            if count == index:
                node = Node(data)
                temp = itr.prev
                temp.next = node
                node.prev = temp
                node.next = itr
                break
            itr = itr.next
            count += 1

    def remove_at(self,index:int):
        itr = self.head
        count = 0
        if count == index:
            self.remove_at_start()
            return
        while itr.next:
            if count == index:
                temp = itr.prev
                temp.next = itr.next
                itr.next.prev = temp
                break
            itr = itr.next
            count += 1

    def print_list(self):
        if self.head is None:
            print("List is empty")
            return

        itr = self.head
        while itr:
            print(itr.data, end=" -> ")
            itr = itr.next
```
