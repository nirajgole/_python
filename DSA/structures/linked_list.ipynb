```python
class Node:
    def __init__(self,data,next=None):
        # each node has data and next pointer refering to next node in list
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_start(self,data):
        node = Node(data,self.head)
        self.head = node

    def insert_at_end(self,data):
        if self.head is None:
            self.head = Node(data,None)
            return

        itr = self.head
        while itr.next:
            itr = itr.next
        itr.next = Node(data,None)

    def remove_at_start(self):
        if self.head is None:
            return
        
        temp = self.head.next
        self.head = temp
        del temp
    
    def remove_at_end(self):
        if self.head is None:
            return
        
        cur = self.head
        while cur:
            if cur.next.next is None:
                break
            cur = cur.next
        temp = cur.next
        cur.next = None
        del temp

            
        # temp, prev.next = prev.next, None
        # prev.next = None
        # del temp

    def insert_at(self,index:int,data:int):
        #try for negative index
        #index starting from 0
        _length = self.get_length()
        if index >= _length or index < 0:
            raise Exception("Index out of range")
        if index == 0:
            self.insert_at_start(data)
        else:
            node = Node(data)
            itr = self.head
            count = 1
            while itr:
                if count == index:
                    temp = itr.next
                    itr.next = node
                    node.next = temp
                    break
                count+=1
                itr = itr.next

    def remove_at(self,index:int):
        _length = self.get_length()
        if index >= _length or index < 0:
            raise Exception("Index out of range")
        if index == 0:
            self.remove_at_start()
        if index == _length-1:
            self.remove_at_end()
        else:
            itr = self.head
            count = 1
            while itr:
                if count == index:
                    temp = itr.next
                    itr.next = itr.next.next
                    del temp
                    break
                count+=1
                itr = itr.next


    def get_length(self):
        if self.head is None:
            return 0
        count = 0
        itr = self.head
        while itr:
            count += 1
            itr = itr.next
        return count

    def print_list(self):
        if self.head is None:
            print("List is empty")
            return

        itr = self.head
        while itr:
            print(itr.data, end=f" -> ")
            itr = itr.next
```
